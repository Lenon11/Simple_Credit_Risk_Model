## Verificación y Validación del Dataset Crediticio

### Introducción
**Contexto**

El presente notebook tiene como objetivo realizar una verificación estructural y validación técnica del dataset crediticio previamente limpiado (credit_dataset_clean.csv).

Antes de construir modelos predictivos o realizar análisis inferenciales, es fundamental garantizar que los datos:

- No contengan valores imposibles o inconsistentes.
- Cumplan con coherencia económica.
- No presenten duplicidades estructurales.
- Mantengan integridad en variables numéricas y categóricas.

**Objetivo del análisis**

Verificar la consistencia estadística, estructural y económica del dataset final, asegurando su idoneidad para modelado crediticio.

```python
import numpy as np
import pandas as pd
```

```python
#Leer credit_dataset_clean.csv
df_model = pd.read_csv("credit_dataset_clean.csv")
```

### Descripción de Variables

Este dataset contiene información financiera y de comportamiento crediticio a nivel mensual por cliente. Cada fila representa un cliente en un mes específico.

#### Identificadores

Estas variables permiten identificar de manera única cada observación.

**Customer_ID**

- Tipo: Identificador único
- Descripción: Código único asignado a cada cliente.
- Función: Permite agrupar observaciones por individuo.
- Uso en modelo: No se utiliza como variable predictora.

**Month**

- Tipo: Variable temporal
- Descripción: Mes correspondiente al registro financiero.
- Función: Permite análisis longitudinal por cliente.
- Uso en modelo: Puede utilizarse para análisis de panel o tendencias.


#### Variable Objetivo

**Credit_Score**

Clasificación crediticia original:

- Poor → Perfil de alto riesgo
- Standard → Perfil intermedio
- Good → Perfil sólido

En modelos tradicionales puede transformarse en variable binaria (ej. Poor vs No Poor).

### **Variables Financieras**

A continuación se describen formalmente las variables numéricas incluidas en el dataset.

#### **Variables Demográficas**

**Age**

- Tipo: Numérica discreta
- Descripción: Edad del cliente.
- Interpretación económica: Puede capturar madurez financiera.

Edad del cliente (18–56 años). Se limpió eliminando valores imposibles y usando la edad consistente por cliente.

#### **Variables de Ingreso**

**Annual_Income**

- Tipo: Numérica continua
- Descripción: Ingreso anual bruto del cliente.
- Importancia: Determina capacidad de pago estructural.

Se aplicó winsorización para eliminar valores extremos irreales.

**Monthly_Inhand_Salary**

Tipo: Numérica continua
Descripción: Ingreso mensual disponible después de deducciones.

Cuando estaba ausente se imputó como:

$$MonthlyInhandSalary = \frac{AnnualIncome}{12}$$

#### **Variables de Endeudamiento**

**Outstanding_Debt**

- Tipo: Numérica continua
- Descripción: Deuda total pendiente.
- Interpretación: Nivel de apalancamiento actual.

**Num_of_Loan**

- Tipo: Numérica discreta
- Descripción: Número total de préstamos activos.
- Interpretación: Intensidad de exposición crediticia.

**Total_EMI_per_month**

- Tipo: Numérica continua
- Descripción: Pago mensual total por créditos (EMI).
- Interpretación: Carga financiera mensual.

#### **Variables de Historial Crediticio**

**Credit_History_Months**

- Tipo: Numérica discreta
- Descripción: Antigüedad crediticia en meses.
- Interpretación: Experiencia en el sistema financiero.

**Num_of_Delayed_Payment**

- Tipo: Numérica discreta
- Descripción: Número de pagos retrasados.
- Interpretación: Indicador directo de riesgo.

**Delay_from_due_date**

- Tipo: Numérica discreta
- Descripción: Días promedio de retraso.
- Interpretación: Severidad de mora.

**Num_Credit_Inquiries**

- Tipo: Numérica discreta
- Descripción: Número de consultas crediticias recientes.
- Interpretación: Señal de posible búsqueda activa de crédito.

#### **Variables de Comportamiento Financiero**

**Credit_Utilization_Ratio**

- Tipo: Numérica continua
- Descripción: Proporción del crédito utilizado respecto al disponible.
- Interpretación: Indicador clave de estrés financiero.

**Monthly_Balance**

- Tipo: Numérica continua
- Descripción: Saldo promedio mensual restante después de pagos.
- Interpretación: Liquidez neta.

**Amount_invested_monthly**

Tipo: Numérica continua
Descripción: Monto invertido mensualmente por el cliente.
Interpretación: Capacidad de ahorro/inversión.

#### **Variables de Infraestructura Financiera**

**Num_Bank_Accounts**

- Tipo: Numérica discreta
- Descripción: Número de cuentas bancarias activas.

**Num_Credit_Card**

- Tipo: Numérica discreta
- Descripción: Número de tarjetas de crédito activas.

#### **Variables de Condiciones Contractuales**

**Interest_Rate**

- Tipo: Numérica continua
- Descripción: Tasa de interés promedio aplicada.

**Changed_Credit_Limit**

- Tipo: Numérica continua
- Descripción: Variación en el límite de crédito.
- Interpretación: Ajustes de confianza crediticia.

### **Variables Categóricas Transformadas para Modelado**

Estas variables fueron codificadas ordinalmente para permitir su uso en modelos estadísticos.

**Credit_Mix_final**

Calidad de mezcla crediticia:

| Valor | Significado                   |
| ----- | ----------------------------- |
| 0     | Bad (mezcla crediticia débil) |
| 1     | Standard (mezcla promedio)    |
| 2     | Good (mezcla saludable)       |

Interpretación: Diversificación del portafolio de crédito.

**Payment_of_Min_Amount_final**

| Valor | Significado       |
| ----- | ----------------- |
| 0     | No paga el mínimo |
| 1     | Sí paga el mínimo |

Indicador de disciplina financiera.

**Spending_Level_final**

| Valor | Significado       |
| ----- | ----------------- |
| 0     | Low (gasto bajo)  |
| 1     | High (gasto alto) |

Clasificación del patrón de gasto.

**Payment_Size_final**

| Valor | Significado             |
| ----- | ----------------------- |
| 0     | Small (pagos pequeños)  |
| 1     | Medium (pagos medianos) |
| 2     | Large (pagos grandes)   |

Tamaño promedio de pagos realizados.

### **Variable Auxiliar**

**Occupation**

- Ocupación principal del cliente.
- Se limpió eliminando valores corruptos y asignando la ocupación consistente por cliente.
- No se incluye directamente en el modelo base tradicional, pero es útil para análisis posteriores.


```python
#Verificación de valores imposibles (numéricas)
num_cols = [
    "Age",
    "Annual_Income",
    "Monthly_Inhand_Salary",
    "Outstanding_Debt",
    "Credit_History_Months",
    "Monthly_Balance",
    "Amount_invested_monthly",
    "Num_of_Loan",
    "Num_of_Delayed_Payment",
    "Changed_Credit_Limit",
    "Interest_Rate",
    "Num_Credit_Inquiries",
    "Total_EMI_per_month",
    "Num_Bank_Accounts",
    "Num_Credit_Card",
    "Delay_from_due_date",
    "Credit_Utilization_Ratio"
]

for col in num_cols:
    print("\n", col)
    print("Min:", df_model[col].min())
    print("Max:", df_model[col].max())
    print("Negativos %:", (df_model[col] < 0).mean())
    print("Inf %:", np.isinf(df_model[col]).mean())
```

```python
#Verificación de duplicados estructurales
print("Duplicados exactos:",
      df_model.duplicated().mean())

print("Duplicados Customer_ID + Month:",
      df_model.duplicated(["Customer_ID", "Month"]).mean())
```

```python
#Verificación de cardinalidad categórica
cat_cols = [
    "Credit_Score",
    "Occupation"
]

for col in cat_cols:
    print("\n", col)
    print("Unique:", df_model[col].nunique())
    print(df_model[col].value_counts().head(10))
```

```python
#Verificación de rangos económicos razonables
print("Debt > Income %:",
      (df_model["Outstanding_Debt"] > df_model["Annual_Income"]).mean())

print("EMI > Monthly Salary %:",
      (df_model["Total_EMI_per_month"] > df_model["Monthly_Inhand_Salary"]).mean())

print("Invested > Monthly Salary %:",
      (df_model["Amount_invested_monthly"] > df_model["Monthly_Inhand_Salary"]).mean())
```

```python
#Verificación de correlación extrema
corr = df_model[num_cols].corr()

high_corr = (
    corr.abs()
        .where(np.triu(np.ones(corr.shape), k=1).astype(bool))
        .stack()
        .sort_values(ascending=False)
)

print(high_corr.head(10))
```

Resultados:

= Los rangos observados son consistentes con comportamiento financiero real.

**Verificación de Duplicados**

Duplicados exactos: 0.0%

**Conclusión**

- Coherencia económica
- Correlaciones razonables
